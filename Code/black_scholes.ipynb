```python
import numpy as np 
import numpy as np 
import math
from scipy.stats import norm
import matplotlib.pyplot as plt
import derivatives as d

N = norm.cdf #normal distrubution function
#assuming no dividend
```

```python
def d1d2(S, K, T, r, sigma):
    d1 = (np.log(S/K) + (r + sigma**2/2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    return d1, d2

def call(d1,d2, K, r, T, S):
    call_value  =  S * N(d1) - K * np.exp(-r*T)* N(d2)
    return call_value

def put(d1,d2, K, r, T, S):
    put_value  =  K*np.exp(-r*T)*N(-d2) - S*N(-d1)
    return put_value
```

```python
underlying = np.arange(0,200,1)
K = 100
T = 1
r = 0.1
sigma = 0.3
d1, d2 = d1d2(underlying, K, T, r, sigma)

call_payff = np.maximum(underlying - K, 0)
plt.plot(underlying, call(d1, d2, K, r, T, underlying))
plt.plot(underlying, call_payff)
```
